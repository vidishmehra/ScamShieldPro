```python
import pandas as pd

file_path = "data/sms_dataset/SMSSpamCollection"

real_df = pd.read_csv(
    file_path,
    sep="\t",
    header=None,
    names=["label", "message"]
)

real_df["result"] = real_df["label"].map({
    "ham": "safe",
    "spam": "scam"
})

print("Dataset loaded!")
print("Total messages:", len(real_df))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    max_features=5000
)

X = vectorizer.fit_transform(real_df["message"])

print("Number of messages:", X.shape[0])
print("Number of features:", X.shape[1])
```

```python
y = real_df["result"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training messages:", X_train.shape[0])
print("Testing messages:", X_test.shape[0])
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

model.fit(X_train, y_train)

print("ScamShield AI model trained successfully!")
```

```python
def calculate_risk_score(message):

    message_lower = message.lower()

    scam_keywords = [
        "urgent",
        "prize",
        "winner",
        "otp",
        "password",
        "click",
        "pay",
        "bank",
        "verify",
        "claim",
        "reward",
        "account"
    ]

    found_keywords = []

    for keyword in scam_keywords:
        if keyword in message_lower:
            found_keywords.append(keyword)

    # Keyword score
    keyword_score = len(found_keywords) * 8

    # ML prediction
    message_vector = vectorizer.transform([message])
    prediction = svm_model.predict(message_vector)[0]

    # Combine ML prediction and keyword score
    if prediction == "scam":
        risk_score = 60 + keyword_score
    else:
        risk_score = keyword_score

    # Keep score between 0 and 100
    risk_score = min(risk_score, 100)

    # Risk level
    if risk_score >= 70:
        risk_level = "HIGH RISK"
    elif risk_score >= 40:
        risk_level = "MEDIUM RISK"
    else:
        risk_level = "LOW RISK"

    return risk_score, risk_level, found_keywords
```

```python
def scamshield_report(message):

    risk_score, risk_level, reasons = calculate_risk_score(message)

    message_vector = vectorizer.transform([message])
    prediction = svm_model.predict(message_vector)[0]

    print("=" * 50)
    print("           SCAMSHIELD PRO REPORT")
    print("=" * 50)

    print("Message:")
    print(message)

    print("\nAI Prediction:")
    print(prediction.upper())

    print("\nRisk Score:")
    print(str(risk_score) + " / 100")

    print("\nRisk Level:")
    print(risk_level)

    print("\nWarning Signs:")

    if reasons:
        for reason in reasons:
            print("⚠️", reason)
    else:
        print("No major warning signs detected.")

    print("=" * 50)

    return {
        "message": message,
        "prediction": prediction,
        "risk_score": risk_score,
        "risk_level": risk_level,
        "warning_signs": reasons
    }
```

```python
from urllib.parse import urlparse

def analyze_url(url):

    suspicious_words = [
        "login",
        "verify",
        "secure",
        "account",
        "update",
        "bank",
        "bonus",
        "prize",
        "claim",
        "free"
    ]

    score = 0
    warning_signs = []

    parsed_url = urlparse(url)

    # Check HTTPS
    if parsed_url.scheme != "https":
        score += 20
        warning_signs.append("Website is not using HTTPS")

    # Check suspicious words
    url_lower = url.lower()

    for word in suspicious_words:
        if word in url_lower:
            score += 10
            warning_signs.append(
                "Suspicious word found: " + word
            )

    # Check URL length
    if len(url) > 75:
        score += 10
        warning_signs.append("Unusually long URL")

    # Check @ symbol
    if "@" in url:
        score += 20
        warning_signs.append(
            "URL contains @ symbol"
        )

    # Limit score
    score = min(score, 100)

    # Risk level
    if score >= 70:
        risk_level = "HIGH RISK"
    elif score >= 40:
        risk_level = "MEDIUM RISK"
    else:
        risk_level = "LOW RISK"

    return score, risk_level, warning_signs
```

```python
def scamshield_analyze(message=None, url=None):

    print("=" * 55)
    print("              SCAMSHIELD PRO")
    print("=" * 55)

    # Message analysis
    if message:
        risk_score, risk_level, reasons = calculate_risk_score(message)

        message_vector = vectorizer.transform([message])
        prediction = svm_model.predict(message_vector)[0]

        print("\n📩 MESSAGE ANALYSIS")
        print("-" * 55)
        print("Message:", message)
        print("AI Prediction:", prediction.upper())
        print("Risk Score:", risk_score, "/ 100")
        print("Risk Level:", risk_level)

        if reasons:
            print("Warning Signs:")
            for reason in reasons:
                print("⚠️", reason)
        else:
            print("✅ No major warning signs detected.")

    # URL analysis
    if url:
        score, level, warnings = analyze_url(url)

        print("\n🔗 URL ANALYSIS")
        print("-" * 55)
        print("URL:", url)
        print("Risk Score:", score, "/ 100")
        print("Risk Level:", level)

        if warnings:
            print("Warning Signs:")
            for warning in warnings:
                print("⚠️", warning)
        else:
            print("✅ No major warning signs detected.")

    print("\n" + "=" * 55)
```

```python
def overall_risk_score(message_score, url_score=None):

    if url_score is None:
        final_score = message_score
    else:
        final_score = (message_score * 0.6) + (url_score * 0.4)

    final_score = round(final_score)

    if final_score >= 70:
        final_level = "HIGH RISK"
    elif final_score >= 40:
        final_level = "MEDIUM RISK"
    else:
        final_level = "LOW RISK"

    return final_score, final_level
```

```python
def final_scamshield_scan(message, url=None):

    print("=" * 60)
    print("                 SCAMSHIELD PRO")
    print("=" * 60)

    # MESSAGE ANALYSIS
    message_score, message_level, message_reasons = calculate_risk_score(message)

    message_vector = vectorizer.transform([message])
    prediction = svm_model.predict(message_vector)[0]

    print("\n📩 MESSAGE ANALYSIS")
    print("-" * 60)
    print("Message:", message)
    print("AI Prediction:", prediction.upper())
    print("Message Risk Score:", message_score, "/ 100")
    print("Message Risk Level:", message_level)

    if message_reasons:
        print("Warning Signs:")
        for reason in message_reasons:
            print("⚠️", reason)
    else:
        print("✅ No major warning signs detected.")

    # URL ANALYSIS
    url_score = None

    if url:
        url_score, url_level, url_warnings = analyze_url(url)

        print("\n🔗 URL ANALYSIS")
        print("-" * 60)
        print("URL:", url)
        print("URL Risk Score:", url_score, "/ 100")
        print("URL Risk Level:", url_level)

        if url_warnings:
            print("Warning Signs:")
            for warning in url_warnings:
                print("⚠️", warning)
        else:
            print("✅ No major URL warning signs detected.")

    # OVERALL SCORE
    final_score, final_level = overall_risk_score(
        message_score,
        url_score
    )

    print("\n🚨 OVERALL SCAMSHIELD RESULT")
    print("-" * 60)
    print("Overall Risk Score:", final_score, "/ 100")
    print("Overall Risk Level:", final_level)

    if final_level == "HIGH RISK":
        print("🚨 WARNING: This content may be dangerous.")
    elif final_level == "MEDIUM RISK":
        print("⚠️ CAUTION: Proceed carefully.")
    else:
        print("✅ No major threat detected.")

    print("=" * 60)
```

```python
def save_scan(
    message,
    url,
    prediction,
    message_score,
    url_score,
    overall_score,
    risk_level
):

    global scan_history

    new_scan = pd.DataFrame([{
        "message": message,
        "url": url,
        "prediction": prediction,
        "message_score": message_score,
        "url_score": url_score,
        "overall_score": overall_score,
        "risk_level": risk_level
    }])

    scan_history = pd.concat(
        [scan_history, new_scan],
        ignore_index=True
    )

    return scan_history
```

```python
# ============================================================
# SCAMSHIELD PRO — AI V2 TRAINING ENGINE
# Stronger NLP + Model Comparison + Cross Validation
# ============================================================

import os
import re
import joblib
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

print("🛡️ ScamShield Pro AI V2")
print("=" * 60)

# ------------------------------------------------------------
# 1. LOAD DATA
# ------------------------------------------------------------

file_path = "data/sms_dataset/SMSSpamCollection"

df = pd.read_csv(
    file_path,
    sep="\t",
    header=None,
    names=["label", "message"]
)

df = df.dropna()

df["label"] = df["label"].map({
    "ham": "safe",
    "spam": "scam"
})

# Remove duplicate messages
df = df.drop_duplicates(subset=["message"]).reset_index(drop=True)

print(f"Dataset size: {len(df)}")
print(df["label"].value_counts())

# ------------------------------------------------------------
# 2. CLEAN TEXT
# ------------------------------------------------------------

def clean_text(text):
    text = str(text).lower()

    # Keep useful security characters such as URLs, numbers,
    # currency symbols and punctuation.
    text = re.sub(r"\s+", " ", text)

    return text.strip()

df["clean_message"] = df["message"].apply(clean_text)

X = df["clean_message"]
y = df["label"]

# ------------------------------------------------------------
# 3. TRAIN / TEST SPLIT
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# ------------------------------------------------------------
# 4. STRONGER TF-IDF
# ------------------------------------------------------------

vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.98,
    sublinear_tf=True,
    max_features=15000
)

X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

print("\nTF-IDF features:", X_train_vec.shape[1])

# ------------------------------------------------------------
# 5. TRAIN THREE MODELS
# ------------------------------------------------------------

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        C=2.0,
        random_state=42
    ),

    "Linear SVM": LinearSVC(
        class_weight="balanced",
        C=1.5,
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=400,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )
}

results = []

trained_models = {}

print("\nTraining models...")
print("-" * 60)

for name, clf in models.items():

    print(f"Training {name}...")

    clf.fit(X_train_vec, y_train)

    predictions = clf.predict(X_test_vec)

    accuracy = accuracy_score(y_test, predictions)
    precision = precision_score(
        y_test,
        predictions,
        pos_label="scam",
        zero_division=0
    )
    recall = recall_score(
        y_test,
        predictions,
        pos_label="scam",
        zero_division=0
    )
    f1 = f1_score(
        y_test,
        predictions,
        pos_label="scam",
        zero_division=0
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Scam Precision": precision,
        "Scam Recall": recall,
        "Scam F1": f1
    })

    trained_models[name] = clf

# ------------------------------------------------------------
# 6. MODEL COMPARISON
# ------------------------------------------------------------

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by=["Scam F1", "Scam Recall"],
    ascending=False
).reset_index(drop=True)

print("\nMODEL PERFORMANCE")
print("=" * 60)

print(
    results_df.to_string(
        index=False,
        formatters={
            "Accuracy": "{:.4f}".format,
            "Scam Precision": "{:.4f}".format,
            "Scam Recall": "{:.4f}".format,
            "Scam F1": "{:.4f}".format
        }
    )
)

# ------------------------------------------------------------
# 7. SELECT BEST MODEL
# ------------------------------------------------------------

best_model_name = results_df.iloc[0]["Model"]
best_model = trained_models[best_model_name]

print("\n🏆 BEST MODEL:", best_model_name)

# ------------------------------------------------------------
# 8. DETAILED EVALUATION
# ------------------------------------------------------------

best_predictions = best_model.predict(X_test_vec)

print("\nDETAILED CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y_test,
        best_predictions,
        digits=4
    )
)

print("\nCONFUSION MATRIX")
print("=" * 60)

cm = confusion_matrix(
    y_test,
    best_predictions,
    labels=["safe", "scam"]
)

print(pd.DataFrame(
    cm,
    index=["Actual Safe", "Actual Scam"],
    columns=["Predicted Safe", "Predicted Scam"]
))

# ------------------------------------------------------------
# 9. CROSS VALIDATION
# ------------------------------------------------------------

print("\n5-FOLD CROSS VALIDATION")
print("=" * 60)

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(
        lowercase=True,
        stop_words="english",
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.98,
        sublinear_tf=True,
        max_features=15000
    )),
    ("model", best_model)
])

cv_scores = cross_val_score(
    cv_pipeline,
    X,
    y,
    cv=cv,
    scoring="f1"
)

print("F1 scores:", np.round(cv_scores, 4))
print("Mean F1:", round(cv_scores.mean(), 4))
print("F1 Std:", round(cv_scores.std(), 4))

# ------------------------------------------------------------
# 10. TRAIN FINAL MODEL ON ALL DATA
# ------------------------------------------------------------

print("\nTraining final model on complete dataset...")

final_pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(
        lowercase=True,
        stop_words="english",
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.98,
        sublinear_tf=True,
        max_features=15000
    )),
    ("model", best_model)
])

final_pipeline.fit(X, y)

# ------------------------------------------------------------
# 11. SAVE MODEL
# ------------------------------------------------------------

os.makedirs("app/models", exist_ok=True)

joblib.dump(
    final_pipeline,
    "app/models/scamshield_ai_v2.pkl"
)

results_df.to_csv(
    "model_comparison_v2.csv",
    index=False
)

# Save evaluation information
evaluation = {
    "Best Model": best_model_name,
    "Test Accuracy": accuracy_score(y_test, best_predictions),
    "Test Scam Precision": precision_score(
        y_test, best_predictions, pos_label="scam", zero_division=0
    ),
    "Test Scam Recall": recall_score(
        y_test, best_predictions, pos_label="scam", zero_division=0
    ),
    "Test Scam F1": f1_score(
        y_test, best_predictions, pos_label="scam", zero_division=0
    ),
    "5-Fold Mean F1": cv_scores.mean(),
    "5-Fold F1 Std": cv_scores.std()
}

evaluation_df = pd.DataFrame(
    [evaluation]
)

evaluation_df.to_csv(
    "scamshield_ai_v2_report.csv",
    index=False
)

# ------------------------------------------------------------
# 12. QUICK REAL-WORLD TESTS
# ------------------------------------------------------------

test_messages = [

    "Congratulations! You won a prize. Click this link now to claim your reward.",

    "URGENT! Your bank account will be suspended. Verify your OTP immediately.",

    "Please send me your password and card details to receive the refund.",

    "Hey, are we still meeting for lunch tomorrow?",

    "Your Amazon order has been delivered successfully.",

    "You have won a cash reward. Pay a small processing fee to claim it."

]

print("\nQUICK THREAT TEST")
print("=" * 60)

for msg in test_messages:

    prediction = final_pipeline.predict([clean_text(msg)])[0]

    print("\nMessage:", msg)
    print("Prediction:", prediction.upper())

print("\n" + "=" * 60)
print("✅ SCAMSHIELD AI V2 TRAINING COMPLETE")
print("=" * 60)

print("\nCreated:")
print("📁 app/models/scamshield_ai_v2.pkl")
print("📊 model_comparison_v2.csv")
print("📊 scamshield_ai_v2_report.csv")

print("\nNext step: connect AI V2 to your existing Flask website.")
```

```python
# ============================================================
# 🛡️ SCAMSHIELD PRO — AI V4 HYBRID SECURITY ENGINE
# ML + SECURITY INTELLIGENCE + HARD-THREAT TESTING
# ============================================================

import os
import re
import json
import joblib
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    make_scorer
)

print("🛡️ ScamShield Pro AI V4")
print("=" * 65)

# ============================================================
# 1. LOAD DATA
# ============================================================

file_path = "data/sms_dataset/SMSSpamCollection"

df = pd.read_csv(
    file_path,
    sep="\t",
    header=None,
    names=["label", "message"]
)

df = df.drop_duplicates()

df["label"] = df["label"].map({
    "ham": "safe",
    "spam": "scam"
})

df["message"] = (
    df["message"]
    .astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

df = df.dropna()

print("Dataset size:", len(df))
print()
print(df["label"].value_counts())

# ============================================================
# 2. TRAIN / TEST SPLIT
# ============================================================

X = df["message"]
y = df["label"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# ============================================================
# 3. WORD + CHARACTER FEATURES
# ============================================================

word_vectorizer = TfidfVectorizer(
    lowercase=True,
    strip_accents="unicode",
    analyzer="word",
    ngram_range=(1, 3),
    min_df=2,
    max_df=0.98,
    sublinear_tf=True,
    max_features=30000
)

char_vectorizer = TfidfVectorizer(
    lowercase=True,
    analyzer="char",
    ngram_range=(3, 5),
    min_df=2,
    max_features=20000,
    sublinear_tf=True
)

# ============================================================
# 4. TRAIN WORD MODEL
# ============================================================

print()
print("Training word-level AI...")

word_model = Pipeline([
    ("tfidf", word_vectorizer),
    ("model", LogisticRegression(
        C=3.0,
        class_weight="balanced",
        max_iter=3000,
        random_state=42
    ))
])

word_model.fit(X_train, y_train)

# ============================================================
# 5. TRAIN CHARACTER MODEL
# ============================================================

print("Training character-level AI...")

char_model = Pipeline([
    ("tfidf", char_vectorizer),
    ("model", LogisticRegression(
        C=2.0,
        class_weight="balanced",
        max_iter=3000,
        random_state=42
    ))
])

char_model.fit(X_train, y_train)

# ============================================================
# 6. SECURITY INTELLIGENCE ENGINE
# ============================================================

security_patterns = {

    "credential_theft": [
        r"\bpassword\b",
        r"\bpasscode\b",
        r"\bpin\b",
        r"\bcvv\b",
        r"\bcard details?\b",
        r"\bcredit card\b",
        r"\bdebit card\b",
        r"\blogin details?\b"
    ],

    "otp_theft": [
        r"\botp\b",
        r"\bone[- ]time password\b",
        r"\bverification code\b",
        r"\bsecurity code\b"
    ],

    "banking": [
        r"\bbank account\b",
        r"\baccount will be suspended\b",
        r"\baccount has been blocked\b",
        r"\baccount has been compromised\b",
        r"\bnet banking\b",
        r"\bupi\b"
    ],

    "kyc_identity": [
        r"\bkyc\b",
        r"\bpan card\b",
        r"\baadhaar\b",
        r"\bidentity verification\b",
        r"\bverify your identity\b"
    ],

    "financial_request": [
        r"\bsend money\b",
        r"\bpay\b",
        r"\bpayment\b",
        r"\btransfer money\b",
        r"\bprocessing fee\b",
        r"\brefund\b",
        r"\bfee\b",
        r"\bdeposit\b"
    ],

    "urgency": [
        r"\burgent\b",
        r"\bimmediately\b",
        r"\bact now\b",
        r"\bwithin \d+ (minutes?|hours?)\b",
        r"\blast warning\b",
        r"\bfinal warning\b",
        r"\bwill be suspended\b",
        r"\bwill be blocked\b",
        r"\bwill be disconnected\b"
    ],

    "prize_reward": [
        r"\byou (have )?won\b",
        r"\bcongratulations\b",
        r"\bprize\b",
        r"\breward\b",
        r"\bfree cash\b",
        r"\bcash reward\b",
        r"\bclaim\b"
    ],

    "suspicious_link": [
        r"https?://",
        r"\bclick (this )?link\b",
        r"\bclick here\b",
        r"\bverify.*link\b",
        r"\blogin.*link\b"
    ],

    "impersonation": [
        r"\bdear customer\b",
        r"\bdear user\b",
        r"\bsecurity team\b",
        r"\bsupport team\b",
        r"\baccount department\b"
    ]
}

# ============================================================
# 7. SECURITY SIGNAL ANALYZER
# ============================================================

def security_analysis(message):

    text = str(message).lower()

    categories = {}
    matched_signals = []

    for category, patterns in security_patterns.items():

        found = []

        for pattern in patterns:

            if re.search(pattern, text):

                found.append(pattern)

        if found:
            categories[category] = len(found)
            matched_signals.append(category)

    # Strong combinations
    strong_attack = False

    credential_combo = (
        "credential_theft" in categories
        and (
            "financial_request" in categories
            or "urgency" in categories
            or "refund" in text
        )
    )

    banking_combo = (
        "banking" in categories
        and (
            "otp_theft" in categories
            or "credential_theft" in categories
            or "urgency" in categories
        )
    )

    prize_combo = (
        "prize_reward" in categories
        and (
            "financial_request" in categories
            or "suspicious_link" in categories
        )
    )

    kyc_combo = (
        "kyc_identity" in categories
        and (
            "urgency" in categories
            or "financial_request" in categories
            or "suspicious_link" in categories
        )
    )

    if credential_combo or banking_combo or prize_combo or kyc_combo:
        strong_attack = True

    return {
        "categories": categories,
        "signals": matched_signals,
        "strong_attack": strong_attack
    }

# ============================================================
# 8. HYBRID PREDICTOR
# ============================================================

def hybrid_predict(message):

    text = str(message)

    # ML predictions
    word_probability = word_model.predict_proba([text])[0]
    char_probability = char_model.predict_proba([text])[0]

    word_classes = list(word_model.classes_)
    char_classes = list(char_model.classes_)

    word_scam_probability = word_probability[
        word_classes.index("scam")
    ]

    char_scam_probability = char_probability[
        char_classes.index("scam")
    ]

    # Combine two ML models
    ml_probability = (
        0.60 * word_scam_probability
        +
        0.40 * char_scam_probability
    )

    # Security intelligence
    intelligence = security_analysis(text)

    categories = intelligence["categories"]

    # Security score
    security_score = 0

    weights = {
        "credential_theft": 35,
        "otp_theft": 30,
        "banking": 20,
        "kyc_identity": 25,
        "financial_request": 25,
        "urgency": 15,
        "prize_reward": 20,
        "suspicious_link": 20,
        "impersonation": 10
    }

    for category in categories:
        security_score += weights.get(category, 0)

    security_score = min(security_score, 100)

    # Convert ML probability to score
    ml_score = ml_probability * 100

    # Hybrid score
    hybrid_score = (
        0.65 * ml_score
        +
        0.35 * security_score
    )

    # Strong security combinations
    if intelligence["strong_attack"]:
        hybrid_score = max(hybrid_score, 82)

    # Explicit credential theft
    if (
        "credential_theft" in categories
        and (
            "financial_request" in categories
            or "otp_theft" in categories
            or "banking" in categories
        )
    ):
        hybrid_score = max(hybrid_score, 90)

    # Explicit prize + payment
    if (
        "prize_reward" in categories
        and "financial_request" in categories
    ):
        hybrid_score = max(hybrid_score, 90)

    hybrid_score = min(round(hybrid_score), 100)

    # Risk classification
    if hybrid_score >= 75:
        prediction = "scam"
        risk_level = "HIGH RISK"

    elif hybrid_score >= 45:
        prediction = "scam"
        risk_level = "MEDIUM RISK"

    else:
        prediction = "safe"
        risk_level = "LOW RISK"

    return {
        "prediction": prediction,
        "risk_score": hybrid_score,
        "risk_level": risk_level,
        "ml_probability": round(ml_probability * 100, 2),
        "security_score": security_score,
        "signals": intelligence["signals"],
        "categories": list(categories.keys())
    }

# ============================================================
# 9. TEST ON HELD-OUT DATA
# ============================================================

print()
print("EVALUATING HYBRID ENGINE...")
print("=" * 65)

hybrid_predictions = []

for message in X_test:

    result = hybrid_predict(message)

    hybrid_predictions.append(
        result["prediction"]
    )

accuracy = accuracy_score(
    y_test,
    hybrid_predictions
)

precision = precision_score(
    y_test,
    hybrid_predictions,
    pos_label="scam"
)

recall = recall_score(
    y_test,
    hybrid_predictions,
    pos_label="scam"
)

f1 = f1_score(
    y_test,
    hybrid_predictions,
    pos_label="scam"
)

print()
print("V4 HYBRID PERFORMANCE")
print("=" * 65)

print("Accuracy:       ", round(accuracy * 100, 2), "%")
print("Scam Precision: ", round(precision * 100, 2), "%")
print("Scam Recall:    ", round(recall * 100, 2), "%")
print("Scam F1:        ", round(f1 * 100, 2), "%")

# ============================================================
# 10. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    hybrid_predictions,
    labels=["safe", "scam"]
)

print()
print("CONFUSION MATRIX")
print("=" * 65)

print("                 Predicted Safe   Predicted Scam")
print(f"Actual Safe          {cm[0][0]:5d}           {cm[0][1]:5d}")
print(f"Actual Scam          {cm[1][0]:5d}           {cm[1][1]:5d}")

# ============================================================
# 11. CLASSIFICATION REPORT
# ============================================================

print()
print("CLASSIFICATION REPORT")
print("=" * 65)

print(
    classification_report(
        y_test,
        hybrid_predictions,
        target_names=["safe", "scam"],
        digits=4
    )
)

# ============================================================
# 12. HARD SECURITY TESTS
# ============================================================

hard_tests = [

    "Congratulations! You won a prize. Click this link now to claim your reward.",

    "URGENT! Your bank account will be suspended. Verify your OTP immediately.",

    "Please send me your password and card details to receive the refund.",

    "Your account has been compromised. Login immediately to secure your account.",

    "Dear customer, your KYC has expired. Update your PAN and bank details immediately.",

    "You have won a cash reward. Pay a small processing fee to claim it.",

    "Your electricity bill is overdue. Pay immediately to avoid disconnection.",

    "Hey, are we still meeting for lunch tomorrow?",

    "Your Amazon order has been delivered successfully.",

    "Thanks for your message. I will call you tomorrow.",

    "Your bank needs your OTP to complete the refund.",

    "Send your CVV and OTP so we can process your refund.",

    "Click the link and login to verify your account.",

    "Your KYC verification failed. Pay ₹10 to reactivate your account.",

    "Your package delivery has been scheduled for tomorrow."
]

print()
print("🔥 HARD SECURITY TEST")
print("=" * 65)

for message in hard_tests:

    result = hybrid_predict(message)

    icon = (
        "🚨"
        if result["prediction"] == "scam"
        else "✅"
    )

    print()
    print("Message:", message)
    print(
        f"Prediction: {icon} "
        f"{result['prediction'].upper()}"
    )
    print(
        f"Risk: {result['risk_score']}/100 "
        f"({result['risk_level']})"
    )
    print(
        f"ML Score: {result['ml_probability']}% | "
        f"Security Score: {result['security_score']}"
    )

    if result["signals"]:
        print(
            "Signals:",
            ", ".join(result["signals"])
        )

# ============================================================
# 13. 5-FOLD CROSS VALIDATION FOR ML CORE
# ============================================================

print()
print("5-FOLD CROSS VALIDATION — ML CORE")
print("=" * 65)

f1_scorer = make_scorer(
    f1_score,
    pos_label="scam"
)

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_scores = cross_val_score(
    word_model,
    X,
    y,
    cv=cv,
    scoring=f1_scorer,
    n_jobs=-1
)

print(
    "F1 scores:",
    np.round(cv_scores, 4)
)

print(
    "Mean F1:",
    round(cv_scores.mean(), 4)
)

print(
    "F1 Std:",
    round(cv_scores.std(), 4)
)

# ============================================================
# 14. SAVE V4 MODELS
# ============================================================

os.makedirs(
    "app/models",
    exist_ok=True
)

joblib.dump(
    word_model,
    "app/models/scamshield_ai_v4_word.pkl"
)

joblib.dump(
    char_model,
    "app/models/scamshield_ai_v4_char.pkl"
)

# Save security configuration
with open(
    "app/models/scamshield_security_rules.json",
    "w"
) as f:

    json.dump(
        {
            "weights": weights,
            "version": "V4"
        },
        f,
        indent=4
    )

# Save performance
v4_report = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Scam Precision",
        "Scam Recall",
        "Scam F1",
        "CV Mean F1",
        "CV F1 Std"
    ],
    "Score": [
        accuracy,
        precision,
        recall,
        f1,
        cv_scores.mean(),
        cv_scores.std()
    ]
})

v4_report.to_csv(
    "scamshield_ai_v4_report.csv",
    index=False
)

# Save hard-test results
hard_results = []

for message in hard_tests:

    result = hybrid_predict(message)

    hard_results.append({
        "Message": message,
        "Prediction": result["prediction"],
        "Risk Score": result["risk_score"],
        "Risk Level": result["risk_level"],
        "ML Score": result["ml_probability"],
        "Security Score": result["security_score"],
        "Signals": ", ".join(result["signals"])
    })

pd.DataFrame(hard_results).to_csv(
    "scamshield_ai_v4_hard_tests.csv",
    index=False
)

# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 65)
print("✅ SCAMSHIELD AI V4 TRAINING COMPLETE")
print("=" * 65)

print()
print("Created:")
print("📁 app/models/scamshield_ai_v4_word.pkl")
print("📁 app/models/scamshield_ai_v4_char.pkl")
print("📁 app/models/scamshield_security_rules.json")
print("📊 scamshield_ai_v4_report.csv")
print("📊 scamshield_ai_v4_hard_tests.csv")

print()
print("🔥 Hybrid architecture ready:")
print("ML + Character AI + Security Intelligence")
```

```python
# ============================================================
# 🛡️ SCAMSHIELD PRO — AI V5
# ADVANCED HYBRID SECURITY ENGINE
# ============================================================

import os
import re
import json
import joblib
import numpy as np
import pandas as pd

print("🛡️ ScamShield Pro AI V5")
print("=" * 70)

# ============================================================
# 1. ADVANCED SECURITY SIGNALS
# ============================================================

v5_patterns = {

    "otp": [
        r"\botp\b",
        r"\bone[- ]time password\b",
        r"\bverification code\b",
        r"\bsecurity code\b"
    ],

    "password": [
        r"\bpassword\b",
        r"\bpasscode\b",
        r"\blogin details?\b"
    ],

    "card_credentials": [
        r"\bcard details?\b",
        r"\bcredit card\b",
        r"\bdebit card\b",
        r"\bcvv\b",
        r"\bcard number\b"
    ],

    "banking": [
        r"\bbank account\b",
        r"\bnet banking\b",
        r"\bupi\b",
        r"\bbank\b"
    ],

    "account_threat": [
        r"\baccount will be suspended\b",
        r"\baccount has been suspended\b",
        r"\baccount will be blocked\b",
        r"\baccount has been blocked\b",
        r"\baccount has been compromised\b",
        r"\baccount will be closed\b"
    ],

    "verification": [
        r"\bverify your account\b",
        r"\bverify your identity\b",
        r"\bverify\b",
        r"\bverification\b"
    ],

    "login": [
        r"\blogin\b",
        r"\bsign in\b",
        r"\bsign-in\b"
    ],

    "link": [
        r"https?://",
        r"\bclick here\b",
        r"\bclick the link\b",
        r"\bclick this link\b",
        r"\bopen this link\b"
    ],

    "payment": [
        r"\bpay\b",
        r"\bpayment\b",
        r"\bsend money\b",
        r"\btransfer money\b",
        r"\bprocessing fee\b",
        r"\bfee\b"
    ],

    "refund": [
        r"\brefund\b",
        r"\bcashback\b",
        r"\bmoney back\b"
    ],

    "kyc": [
        r"\bkyc\b",
        r"\bpan card\b",
        r"\bpan\b",
        r"\baadhaar\b"
    ],

    "urgency": [
        r"\burgent\b",
        r"\bimmediately\b",
        r"\bact now\b",
        r"\bnow\b",
        r"\bas soon as possible\b",
        r"\blast warning\b",
        r"\bfinal warning\b"
    ],

    "prize": [
        r"\byou have won\b",
        r"\byou won\b",
        r"\bcongratulations\b",
        r"\bprize\b",
        r"\breward\b",
        r"\bfree cash\b",
        r"\bcash reward\b"
    ],

    "delivery": [
        r"\border\b",
        r"\bpackage\b",
        r"\bdelivery\b",
        r"\bdelivered\b",
        r"\bshipment\b"
    ]
}

# ============================================================
# 2. SIGNAL DETECTOR
# ============================================================

def detect_v5_signals(message):

    text = str(message).lower()

    found = {}

    for category, patterns in v5_patterns.items():

        count = 0

        for pattern in patterns:

            if re.search(pattern, text):
                count += 1

        if count > 0:
            found[category] = count

    return found


# ============================================================
# 3. ADVANCED SECURITY SCORING
# ============================================================

def v5_security_score(message):

    text = str(message).lower()

    signals = detect_v5_signals(message)

    score = 0
    reasons = []

    # --------------------------------------------------------
    # DIRECT HIGH-RISK CREDENTIAL REQUESTS
    # --------------------------------------------------------

    if "otp" in signals:
        score += 35
        reasons.append("OTP/security-code request")

    if "password" in signals:
        score += 35
        reasons.append("Password/login credential request")

    if "card_credentials" in signals:
        score += 40
        reasons.append("Card/CVV information request")

    # --------------------------------------------------------
    # BANKING
    # --------------------------------------------------------

    if "banking" in signals:
        score += 15
        reasons.append("Banking-related content")

    if "account_threat" in signals:
        score += 30
        reasons.append("Account threat or suspension language")

    # --------------------------------------------------------
    # VERIFICATION + LOGIN + LINK
    # --------------------------------------------------------

    if "verification" in signals:
        score += 15
        reasons.append("Verification request")

    if "login" in signals:
        score += 15
        reasons.append("Login request")

    if "link" in signals:
        score += 20
        reasons.append("Link/click action")

    # Strong phishing combination
    if (
        "verification" in signals
        and "login" in signals
        and "link" in signals
    ):
        score += 30
        reasons.append(
            "Phishing combination: link + login + verification"
        )

    # --------------------------------------------------------
    # FINANCIAL
    # --------------------------------------------------------

    if "payment" in signals:
        score += 25
        reasons.append("Payment/money request")

    if "refund" in signals:
        score += 15
        reasons.append("Refund-related request")

    # Refund + credential request
    if (
        "refund" in signals
        and (
            "otp" in signals
            or "password" in signals
            or "card_credentials" in signals
        )
    ):
        score += 35
        reasons.append(
            "Refund combined with credential request"
        )

    # --------------------------------------------------------
    # KYC
    # --------------------------------------------------------

    if "kyc" in signals:
        score += 25
        reasons.append("KYC/identity information request")

    if (
        "kyc" in signals
        and (
            "payment" in signals
            or "link" in signals
            or "urgency" in signals
        )
    ):
        score += 30
        reasons.append(
            "KYC combined with suspicious action"
        )

    # --------------------------------------------------------
    # URGENCY
    # --------------------------------------------------------

    if "urgency" in signals:
        score += 15
        reasons.append("Urgency or pressure language")

    # --------------------------------------------------------
    # PRIZE
    # --------------------------------------------------------

    if "prize" in signals:
        score += 25
        reasons.append("Prize/reward claim")

    if (
        "prize" in signals
        and (
            "payment" in signals
            or "link" in signals
        )
    ):
        score += 30
        reasons.append(
            "Prize combined with payment/link"
        )

    # --------------------------------------------------------
    # DELIVERY SAFETY SIGNAL
    # --------------------------------------------------------

    # Normal delivery messages should not be treated as scams
    # simply because they contain words like order/package.
    if "delivery" in signals:

        if not any([
            "otp" in signals,
            "password" in signals,
            "card_credentials" in signals,
            "account_threat" in signals,
            "kyc" in signals,
            "prize" in signals
        ]):

            score -= 15

    score = max(0, min(score, 100))

    return score, signals, reasons


# ============================================================
# 4. HYBRID V5 PREDICTOR
# ============================================================

def v5_predict(message):

    text = str(message)

    # Word model probability
    word_probability = word_model.predict_proba([text])[0]

    word_classes = list(word_model.classes_)

    ml_score = (
        word_probability[
            word_classes.index("scam")
        ] * 100
    )

    # Security intelligence
    security_score, signals, reasons = v5_security_score(text)

    # --------------------------------------------------------
    # HYBRID SCORE
    # --------------------------------------------------------

    final_score = (
        0.55 * ml_score
        +
        0.45 * security_score
    )

    # --------------------------------------------------------
    # HARD SECURITY OVERRIDES
    # --------------------------------------------------------

    # OTP + bank/account
    if (
        "otp" in signals
        and "banking" in signals
    ):
        final_score = max(final_score, 85)

    # OTP + refund
    if (
        "otp" in signals
        and "refund" in signals
    ):
        final_score = max(final_score, 88)

    # Card/CVV + payment/refund
    if (
        "card_credentials" in signals
        and (
            "payment" in signals
            or "refund" in signals
        )
    ):
        final_score = max(final_score, 92)

    # Password + payment/refund
    if (
        "password" in signals
        and (
            "payment" in signals
            or "refund" in signals
        )
    ):
        final_score = max(final_score, 90)

    # Link + login + verification
    if (
        "link" in signals
        and "login" in signals
        and "verification" in signals
    ):
        final_score = max(final_score, 85)

    # KYC + payment
    if (
        "kyc" in signals
        and "payment" in signals
    ):
        final_score = max(final_score, 85)

    # Prize + payment
    if (
        "prize" in signals
        and "payment" in signals
    ):
        final_score = max(final_score, 90)

    final_score = round(
        max(0, min(final_score, 100))
    )

    # --------------------------------------------------------
    # FINAL CLASSIFICATION
    # --------------------------------------------------------

    if final_score >= 75:

        prediction = "scam"
        risk_level = "HIGH RISK"

    elif final_score >= 45:

        prediction = "scam"
        risk_level = "MEDIUM RISK"

    else:

        prediction = "safe"
        risk_level = "LOW RISK"

    return {
        "prediction": prediction,
        "risk_score": final_score,
        "risk_level": risk_level,
        "ml_score": round(ml_score, 2),
        "security_score": security_score,
        "signals": list(signals.keys()),
        "reasons": reasons
    }


# ============================================================
# 5. HARD SECURITY TEST
# ============================================================

hard_tests_v5 = [

    "Congratulations! You won a prize. Click this link now to claim your reward.",

    "URGENT! Your bank account will be suspended. Verify your OTP immediately.",

    "Please send me your password and card details to receive the refund.",

    "Your account has been compromised. Login immediately to secure your account.",

    "Dear customer, your KYC has expired. Update your PAN and bank details immediately.",

    "You have won a cash reward. Pay a small processing fee to claim it.",

    "Your electricity bill is overdue. Pay immediately to avoid disconnection.",

    "Hey, are we still meeting for lunch tomorrow?",

    "Your Amazon order has been delivered successfully.",

    "Thanks for your message. I will call you tomorrow.",

    "Your bank needs your OTP to complete the refund.",

    "Send your CVV and OTP so we can process your refund.",

    "Click the link and login to verify your account.",

    "Your KYC verification failed. Pay ₹10 to reactivate your account.",

    "Your package delivery has been scheduled for tomorrow.",

    "Click here to verify your bank account immediately.",

    "Your account will be blocked unless you login and verify now.",

    "Please provide your OTP to receive your refund.",

    "Congratulations, you have won ₹50,000. Pay ₹500 to release the prize.",

    "Your package has been shipped and will arrive tomorrow."
]

print()
print("🔥 V5 HARD SECURITY TEST")
print("=" * 70)

v5_results = []

for message in hard_tests_v5:

    result = v5_predict(message)

    icon = (
        "🚨"
        if result["prediction"] == "scam"
        else "✅"
    )

    print()
    print("Message:", message)
    print(
        "Prediction:",
        icon,
        result["prediction"].upper()
    )
    print(
        "Risk:",
        result["risk_score"],
        "/100",
        "(" + result["risk_level"] + ")"
    )
    print(
        "ML Score:",
        result["ml_score"],
        "| Security Score:",
        result["security_score"]
    )

    if result["signals"]:
        print(
            "Signals:",
            ", ".join(result["signals"])
        )

    v5_results.append({
        "Message": message,
        "Prediction": result["prediction"],
        "Risk Score": result["risk_score"],
        "Risk Level": result["risk_level"],
        "ML Score": result["ml_score"],
        "Security Score": result["security_score"],
        "Signals": ", ".join(result["signals"]),
        "Reasons": " | ".join(result["reasons"])
    })


# ============================================================
# 6. SAVE V5 ENGINE
# ============================================================

os.makedirs(
    "app/models",
    exist_ok=True
)

joblib.dump(
    word_model,
    "app/models/scamshield_ai_v5_word.pkl"
)

with open(
    "app/models/scamshield_v5_rules.json",
    "w"
) as f:

    json.dump(
        {
            "version": "V5",
            "description": "Hybrid ML + Security Intelligence Engine"
        },
        f,
        indent=4
    )

pd.DataFrame(v5_results).to_csv(
    "scamshield_ai_v5_hard_tests.csv",
    index=False
)

print()
print("=" * 70)
print("✅ SCAMSHIELD AI V5 READY")
print("=" * 70)

print()
print("Created:")
print("📁 app/models/scamshield_ai_v5_word.pkl")
print("📁 app/models/scamshield_v5_rules.json")
print("📊 scamshield_ai_v5_hard_tests.csv")
```

```python
from pathlib import Path
import app

p = Path(app.__file__).resolve()
s = p.read_text()

target = '    risk_score = round(min(100, max(0, risk_score)))'

if target not in s:
    raise Exception("Could not find the risk-score line in app.py")

fix = r'''
    # LEGITIMATE PAYMENT / CREDIT / TRANSFER CONTEXT
    benign_transaction_patterns = [
        r"\bpayment .*successfully processed\b",
        r"\bpayment was successful\b",
        r"\bpayment has been successful\b",
        r"\btransaction was successful\b",
        r"\btransaction has been completed\b",
        r"\bmoney has been credited\b",
        r"\bamount has been credited\b",
        r"\bfriend sent you\b",
        r"\bhas been credited to your account\b",
        r"\bcredited to your account\b"
    ]

    # Strong account-threat language should outweigh generic banking terms.
    strong_account_threat_patterns = [
        r"\baccount .*blocked\b",
        r"\baccount .*suspended\b",
        r"\baccount .*will be blocked\b",
        r"\baccount .*will be suspended\b",
        r"\bunless you verify\b",
        r"\bverify your identity\b.*\blink\b",
        r"\bclick .*link\b.*\bverify\b"
    ]

    benign_transaction_context = any(
        re.search(pattern, message_lower)
        for pattern in benign_transaction_patterns
    )

    strong_account_threat = any(
        re.search(pattern, message_lower)
        for pattern in strong_account_threat_patterns
    )

    if benign_transaction_context and not strong_scam_context:
        risk_score = min(risk_score, 30)
        detected.pop("Banking", None)
        detected.pop("Payment", None)

    if strong_account_threat:
        risk_score = max(risk_score, 75)
'''

s = s.replace(target, fix + "\n" + target, 1)
p.write_text(s)

print("✅ Transaction false-positive protection added.")
print("✅ Account-threat detection strengthened.")
print("File:", p)
```

```python
import pandas as pd
from pathlib import Path

file_path = Path("data/sms_dataset/SMSSpamCollection")

real_df = pd.read_csv(
    file_path,
    sep="\t",
    header=None,
    names=["label", "message"]
)

real_df["result"] = real_df["label"].map({
    "ham": "safe",
    "spam": "scam"
})

print("✅ Dataset loaded")
print("Total messages:", len(real_df))
print("Safe messages:", (real_df["result"] == "safe").sum())
print("Scam messages:", (real_df["result"] == "scam").sum())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Recreate the same train/test split used during model development
X_train_eval, X_test_eval, y_train_eval, y_test_eval = train_test_split(
    real_df["message"],
    real_df["result"],
    test_size=0.20,
    random_state=42,
    stratify=real_df["result"]
)

# Predict using the current ScamShield ML model
predictions = app.v5_word_model.predict(X_test_eval)

accuracy = accuracy_score(y_test_eval, predictions)
precision = precision_score(y_test_eval, predictions, pos_label="scam")
recall = recall_score(y_test_eval, predictions, pos_label="scam")
f1 = f1_score(y_test_eval, predictions, pos_label="scam")
cm = confusion_matrix(
    y_test_eval,
    predictions,
    labels=["safe", "scam"]
)

print("===== SCAMSHIELD ML EVALUATION =====")
print(f"Accuracy : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")

print("\nConfusion Matrix")
print("                 Predicted")
print("                 Safe   Scam")
print(f"Actual Safe      {cm[0][0]:4d}   {cm[0][1]:4d}")
print(f"Actual Scam      {cm[1][0]:4d}   {cm[1][1]:4d}")
```
